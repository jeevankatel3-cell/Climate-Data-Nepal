```python
# Raster Preparation for Final Models 
# Refer to notes for all variables and resampling requirement

import os
import rasterio
import numpy as np
import pandas as pd
from rasterio.enums import Resampling
import rioxarray as rxr
import xarray as xr
import glob

#Date filter for preanalysis - 2001 start - 2005 end
start_date = 20010101
end_date = 20051231

# Input and processed variables used as is from the original dir - Kriged ERA5L, MODIS LST and EVI rasters, DEM and Fetch Variables 
# Variables that are resampled - ERA5L, CHIRPS, HAR, PERSIANN; ERA5L weather variables 

# Define input and output directories
ERA5L_dir = r"..\Data\Datasets\ERA5L_prec"
CHIRPS_dir = r"..\Data\Datasets\CHIRPS_prec"
HAR_dir = r"..\Data\Datasets\HAR_prec"
PERSIANN_dir = r"..\Data\Datasets\PERSIANN_prec"
ERA5L_wind_speed_dir = r"..\Data\Datasets\ERA5L_2_wind_speed"
ERA5L_total_evaporation_dir = r"..\Data\Datasets\ERA5L_2_total_evaporation"
ERA5L_surface_pressure_dir = r"..\Data\Datasets\ERA5L_2_surface_pressure"
ERA5L_soil_moisture_dir = r"..\Data\Datasets\ERA5L_2_soil_moisture"
ERA5L_dewpoint_temperature_dir = r"..\Data\Datasets\ERA5L_2_dewpoint_temperature"
MODIS_EVI_dir = r"..\Data\Datasets\MODIS_2_EVI"


# Temp directory to store resampled rasters
temp_dir = r"..\Data\Processed\Classification_Regression\Temp_Rasters"
os.makedirs(temp_dir, exist_ok=True)

#Template WGS84 raster for resampling
wgs84_raster_path = r"..\Data\Datasets\DEM_1km\DEM_1km_elevation_max.tif"
```

```python
# Helper function to resample rasters to match the WGS84 template
def clip_and_resample_raster(input_raster, sample_raster = rxr.open_rasterio(wgs84_raster_path)):
    resampled_raster = input_raster.rio.reproject_match(sample_raster, resampling= Resampling.bilinear)
    return resampled_raster
```

```python
# Resample ERA5L rasters and save to temp directory
era5L_files = sorted(glob.glob(os.path.join(ERA5L_dir, "*.tif")))
era5L_dates = [pd.to_datetime(f[-12:-4], format='%Y%m%d') for f in era5L_files]

for file, date in zip(era5L_files, era5L_dates):
    if date >= pd.to_datetime(start_date, format='%Y%m%d') and date <= pd.to_datetime(end_date, format='%Y%m%d'):
        input_raster = rxr.open_rasterio(file)
        resampled_raster = clip_and_resample_raster(input_raster)
        resampled_dir = os.path.join(temp_dir, "ERA5L_resampled")
        os.makedirs(resampled_dir, exist_ok=True)
        output_path = os.path.join(resampled_dir, f"ERA5L_prec_{date.strftime('%Y%m%d')}_resampled.tif")
        resampled_raster.rio.to_raster(output_path)
    else:
        continue    
```

```python
# For modis evi (16 day) composite we perform temporal resampling using linear interpolation then to sample raster to get daily values
import bisect
import shutil

modis_evi_files = sorted(glob.glob(os.path.join(MODIS_EVI_dir, "*.tif")))
modis_evi_dates = [pd.to_datetime(f[-12:-4], format='%Y%m%d') for f in modis_evi_files]

start_dt = pd.to_datetime(start_date, format='%Y%m%d')
end_dt   = pd.to_datetime(end_date,   format='%Y%m%d')

# Include one composite before start and one after end for clean interpolation
all_dates_series = pd.Series(modis_evi_dates)
buffer_mask = (
    (all_dates_series >= start_dt - pd.Timedelta(days=32)) &
    (all_dates_series <= end_dt   + pd.Timedelta(days=32))
)
sel_files = [f for f, m in zip(modis_evi_files, buffer_mask) if m]
sel_dates = [d for d, m in zip(modis_evi_dates, buffer_mask) if m]

# 1. Resample and save 16-day composites to a temporary intermediate directory to save RAM
temp_intermediate_dir = os.path.join(temp_dir, "MODIS_EVI_intermediate")
os.makedirs(temp_intermediate_dir, exist_ok=True)

intermediate_files = []
print(f"Step 1: Resampling {len(sel_files)} 16-day MODIS composites and saving intermediate files...")
for file in sel_files:
    basename = os.path.basename(file)
    intermediate_path = os.path.join(temp_intermediate_dir, basename)
    intermediate_files.append(intermediate_path)
    
    # Skip if it already exists (useful if the process crashed midway)
    if not os.path.exists(intermediate_path):
        raster = rxr.open_rasterio(file)
        resampled = clip_and_resample_raster(raster)
        if "band" in resampled.dims and resampled.sizes["band"] == 1:
            resampled = resampled.squeeze("band", drop=True)
        resampled.rio.to_raster(intermediate_path)
        raster.close()

# 2. Interpolate daily values exclusively reading the 2 needed intermediate files to avoid memory crash
print("Step 2: Interpolating daily values from intermediate files...")
resampled_dir = os.path.join(temp_dir, "MODIS_EVI_resampled")
os.makedirs(resampled_dir, exist_ok=True)

daily_dates = pd.date_range(start_dt, end_dt, freq="D")

for i, date in enumerate(daily_dates):
    idx = bisect.bisect_right(sel_dates, date)
    
    if idx == 0:
        target_file = intermediate_files[0]
        do_interp = False
    elif idx == len(sel_dates) and date > sel_dates[-1]:
        target_file = intermediate_files[-1]
        do_interp = False
    elif sel_dates[idx-1] == date:
        target_file = intermediate_files[idx-1]
        do_interp = False
    else:
        do_interp = True
        
    if not do_interp:
        daily_slice = rxr.open_rasterio(target_file)
        if "band" in daily_slice.dims and daily_slice.sizes["band"] == 1:
            daily_slice = daily_slice.squeeze("band", drop=True)
            
        output_path = os.path.join(resampled_dir, f"MODIS_EVI_{date.strftime('%Y%m%d')}_resampled.tif")
        daily_slice.rio.to_raster(output_path)
        daily_slice.close()
    else:
        d1 = sel_dates[idx-1]
        d2 = sel_dates[idx]
        f1 = intermediate_files[idx-1]
        f2 = intermediate_files[idx]
        
        r1 = rxr.open_rasterio(f1)
        r2 = rxr.open_rasterio(f2)
        
        if "band" in r1.dims and r1.sizes["band"] == 1:
            r1 = r1.squeeze("band", drop=True)
        if "band" in r2.dims and r2.sizes["band"] == 1:
            r2 = r2.squeeze("band", drop=True)
            
        weight = (date - d1).days / (d2 - d1).days
        nodata = r1.rio.nodata
        
        if nodata is not None:
            r1_val = r1.where(r1 != nodata)
            r2_val = r2.where(r2 != nodata)
        else:
            r1_val = r1
            r2_val = r2
            
        daily_slice = r1_val + (r2_val - r1_val) * weight
        
        if nodata is not None:
            daily_slice = daily_slice.fillna(nodata)
            
        daily_slice = daily_slice.rio.write_crs(r1.rio.crs)
        daily_slice = daily_slice.rio.write_transform(r1.rio.transform())
        if nodata is not None:
            daily_slice = daily_slice.rio.write_nodata(nodata)
            
        output_path = os.path.join(resampled_dir, f"MODIS_EVI_{date.strftime('%Y%m%d')}_resampled.tif")
        daily_slice.rio.to_raster(output_path)
        
        r1.close()
        r2.close()

print("Step 3: Cleaning up intermediate files...")
try:
    shutil.rmtree(temp_intermediate_dir)
except Exception as e:
    print(f"Warning: Could not remove temporary directory: {e}")

print("Temporal resampling of MODIS EVI complete.")
```
