```python
import xarray as xr 
import rioxarray as rxr 
import rasterio
from rasterio.enums import Resampling

template_wgs_1km = r"..\..\Data\Datasets\DEM_1km\DEM_1km_elevation_max.tif"
template_utm_1km = r"..\..\Data\Datasets\template_raster_UTM45N.tif"

wgs_1km_rxr = rxr.open_rasterio(template_wgs_1km)
utm_1km_rxr = rxr.open_rasterio(template_utm_1km)
```

```python
def change_resolution(raster, km_resolution):
    upscale_factor = 1/km_resolution
    new_width = raster.rio.width * upscale_factor
    new_height = raster.rio.height * upscale_factor
    up_sampled = raster.rio.reproject(raster.rio.crs, shape=(int(new_height), int(new_width)), resampling=Resampling.bilinear)
    return up_sampled
```

```python
wgs_1km_rxr.rio.to_raster(r"..\..\Data\Datasets\TemplateRasters\template_raster_WGS_1km.tif")

wgs_1_5km_rxr = change_resolution(wgs_1km_rxr, 1.5)
wgs_1_5km_rxr.rio.to_raster(r"..\..\Data\Datasets\TemplateRasters\template_raster_WGS_1_5km.tif")

wgs_2km_rxr = change_resolution(wgs_1km_rxr, 2)
wgs_2km_rxr.rio.to_raster(r"..\..\Data\Datasets\TemplateRasters\template_raster_WGS_2km.tif")

wgs_2_5km_rxr = change_resolution(wgs_1km_rxr, 2.5)
wgs_2_5km_rxr.rio.to_raster(r"..\..\Data\Datasets\TemplateRasters\template_raster_WGS_2_5km.tif")

wgs_3km_rxr = change_resolution(wgs_1km_rxr, 3)
wgs_3km_rxr.rio.to_raster(r"..\..\Data\Datasets\TemplateRasters\template_raster_WGS_3km.tif")

wgs_3_5km_rxr = change_resolution(wgs_1km_rxr, 3.5)
wgs_3_5km_rxr.rio.to_raster(r"..\..\Data\Datasets\TemplateRasters\template_raster_WGS_3_5km.tif")

wgs_4km_rxr = change_resolution(wgs_1km_rxr, 4)
wgs_4km_rxr.rio.to_raster(r"..\..\Data\Datasets\TemplateRasters\template_raster_WGS_4km.tif")

wgs_4_5km_rxr = change_resolution(wgs_1km_rxr, 4.5)
wgs_4_5km_rxr.rio.to_raster(r"..\..\Data\Datasets\TemplateRasters\template_raster_WGS_4_5km.tif")

wgs_5km_rxr = change_resolution(wgs_1km_rxr, 5)
wgs_5km_rxr.rio.to_raster(r"..\..\Data\Datasets\TemplateRasters\template_raster_WGS_5km.tif")

wgs_5_5km_rxr = change_resolution(wgs_1km_rxr, 5.5)
wgs_5_5km_rxr.rio.to_raster(r"..\..\Data\Datasets\TemplateRasters\template_raster_WGS_5_5km.tif")

wgs_6km_rxr = change_resolution(wgs_1km_rxr, 6)
wgs_6km_rxr.rio.to_raster(r"..\..\Data\Datasets\TemplateRasters\template_raster_WGS_6km.tif")
```
